Write a SQL query using a chain of Common Table Expressions (CTEs) to identify movies whose average rating is at least 0.5 points higher than the average rating of their genre. Only include movies that have received at least 30 ratings.

```python
WITH genre_avg AS (
    SELECT
        genres.genre_id,
        genres.genre_name,
        AVG(ratings.rating) AS genre_avg_rating
    FROM ratings
    INNER JOIN movie_genres
        ON ratings.movie_id = movie_genres.movie_id
    INNER JOIN genres
        ON movie_genres.genre_id = genres.genre_id
    GROUP BY genres.genre_id, genres.genre_name
)
```

```python
WITH movie_avg AS (
    SELECT
        movies.movie_id,
        movies.movie_title,
        AVG(ratings.rating) AS movie_avg_rating,
        COUNT(*) AS total_ratings
    FROM movies
    INNER JOIN ratings
        ON movies.movie_id = ratings.movie_id
    GROUP BY movies.movie_id, movies.movie_title
    HAVING COUNT(*) >= 30
)
```

```python
movie_comparison AS (
    SELECT
        movie_avg.movie_title,
        genres.genre_name,
        movie_avg.movie_avg_rating,
        genre_avg.genre_avg_rating,
        movie_avg.total_ratings
    FROM movie_avg
    INNER JOIN movie_genres
        ON movie_avg.movie_id = movie_genres.movie_id
    INNER JOIN genres
        ON movie_genres.genre_id = genres.genre_id
    INNER JOIN genre_avg
        ON genres.genre_id = genre_avg.genre_id
)
```

```python
SELECT
    movie_title,
    genre_name,
    ROUND(movie_avg_rating, 2) AS movie_avg_rating,
    ROUND(genre_avg_rating, 2) AS genre_avg_rating,
    total_ratings
FROM movie_comparison
WHERE movie_avg_rating >= genre_avg_rating + 0.5
ORDER BY movie_avg_rating DESC, movie_title, genre_name;
```
